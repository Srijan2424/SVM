```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import numpy as np 
import sklearn
```

```python
# lets create synthetic data points
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=2,n_redundant=0)
```

```python
X
```

```python
y
```

```python
pd.DataFrame(X)
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0],y=pd.DataFrame(X)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
svc=SVC(kernel="linear")
```

```python
svc.fit(X_train,y_train)
```

```python
# Make a Prediction 
pred_val=svc.predict(X_test)
```

```python
# check the accuracy of the model 
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_test,pred_val))
```

```python
print(confusion_matrix(y_test,pred_val))
```

```python
y_pred_rbf=rbf.predict(X_test)
```

```python
# the performance assesment
print(classification_report(y_test,y_pred_rbf))
```

```python
# sigmoid
sigmoid=SVC(kernel="sigmoid")
sigmoid.fit(X_train,y_train)
y_pred_sig=sigmoid.predict(X_test)
# measuring the accuracy
print(classification_report(y_test,y_pred_sig))
```

# SVM Kernels and Indepth Intuition and Pranctical Explanation

```python
x=np.linspace(-5.0,5.0,100)
y=np.sqrt(10**2 - x**2)
y=np.hstack([y,-y])
x=np.hstack([x,-x])
```

```python
x1=np.linspace(-5.0,5.0,100)
y1=np.sqrt(5**2 - x1**2)
y1=np.hstack([y1,-y1])
x1=np.hstack([x1,-x1])
```

```python
plt.scatter(y,x)
plt.scatter(y1,x1)
```

```python
df1=pd.DataFrame(np.vstack([y,x]).T,columns=["X1","X2"])
df1['Y']=0
df2=pd.DataFrame(np.vstack([y1,x1]).T,columns=["X1","X2"])
df2['Y']=1
df=pd.merge(df1,df2,how="outer")
df.head(5)
```

```python
#  independent and dependent features
X=df.iloc[:,:2]
y=df.Y
```

```python
y
```

```python
X
```

```python
# split the data into train test split 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
# dataset 
import seaborn as sns
df=sns.load_dataset("tips")
```

```python
# independent and dependednt features
X=df[[ 'tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df["total_bill"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
X_train
```

```python
# Feature Encoding(Label Encoding and One Hot Encoding)
```

```python
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
X_train["sex"]=le1.fit_transform(X_train["sex"])
X_train["smoker"]=le2.fit_transform(X_train["smoker"])
X_train["time"]=le3.fit_transform(X_train["time"])
```

```python
X_train
```

```python
X_test["sex"]=le1.fit(X_test["sex"])
X_test["smoker"]=le2.fit(X_test["smoker"])
X_test["time"]=le3.fit(X_test["time"])
```

```python
# onehot encoder ---- column transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[("onehot",OneHotEncoder(drop="first"),[3])],remainder='passthrough')
```

```python
import sys
import numpy as np
np.set_printoptions(threshold=sys.maxsize)
X_train=ct.fit_transform(X_train)
```

```python
X_test=ct.fit(X_test)
```

```python
# SVR -- Support Vector Reggresor
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_predict_svr=svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_predict_svr))
```

```python
print(mean_absolute_error(y_test,y_predict_svr))
```
